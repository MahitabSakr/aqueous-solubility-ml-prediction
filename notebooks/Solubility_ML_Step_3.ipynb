## Scaling

```python
from sklearn.preprocessing import StandardScaler

# Create an instance of the scaler
scaler = StandardScaler()

# Fit on training data
scaler.fit(X_train)

# Transform both training and test data
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(x_test)
X_val_scaled = scaler.transform(x_val)
X_train = X_train_scaled
x_test = X_test_scaled
x_val = X_val_scaled
```

## Model Building 

### Linear Regression

```python
#Training the model

from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
#Applying the model to make a prediction

y_lr_train_pred = lr.predict(X_train)
y_lr_val_pred = lr.predict(x_val)
y_lr_test_pred = lr.predict(x_test)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_percentage_error

# Initialize lists to store metrics for each run
train_mse, train_r2, train_rmse, train_mape = [], [], [], []
test_mse, test_r2, test_rmse, test_mape = [], [], [], []

# Run the model training and evaluation 1000 times
for _ in range(1000):
    # Training the model
    et = ExtraTreesRegressor()
    et.fit(X_train, y_train)

    # Applying the model to make a prediction
    y_et_train_pred = et.predict(X_train)
    y_et_test_pred = et.predict(x_test)

    # Evaluate model performance
    train_mse.append(mean_squared_error(y_train, y_et_train_pred))
    train_r2.append(r2_score(y_train, y_et_train_pred))
    train_rmse.append(np.sqrt(train_mse[-1]))
    train_mape.append(mean_absolute_percentage_error(y_train, y_et_train_pred))

    test_mse.append(mean_squared_error(y_test, y_et_test_pred))
    test_r2.append(r2_score(y_test, y_et_test_pred))
    test_rmse.append(np.sqrt(test_mse[-1]))
    test_mape.append(mean_absolute_percentage_error(y_test, y_et_test_pred))

# Calculate means and standard deviations
metrics = {
    "Metric": ["MSE", "R2", "RMSE", "MAPE"],
    "Train Mean": [np.mean(train_mse), np.mean(train_r2), np.mean(train_rmse), np.mean(train_mape)],
    "Train Std": [np.std(train_mse), np.std(train_r2), np.std(train_rmse), np.std(train_mape)],
    "Test Mean": [np.mean(test_mse), np.mean(test_r2), np.mean(test_rmse), np.mean(test_mape)],
    "Test Std": [np.std(test_mse), np.std(test_r2), np.std(test_rmse), np.std(test_mape)]
}

# Convert to DataFrame for a nice table format
results_df = pd.DataFrame(metrics)

# Print the results table
print(results_df)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import norm

# Number of samples
n = len(results_df)

# Calculate the 95% confidence interval for each metric
confidence_level = 0.95
z_score = norm.ppf((1 + confidence_level) / 2)  # Z-score for 95% CI

# Add CI columns to the DataFrame
for metric in ['Train', 'Test']:
    mean = results_df[f'{metric} Mean']
    std = results_df[f'{metric} Std']
    ci_margin = z_score * std / np.sqrt(n)
   
    results_df[f'{metric} 95% CI Lower'] = mean - ci_margin
    results_df[f'{metric} 95% CI Upper'] = mean + ci_margin

# Display the updated DataFrame
print(results_df)
```

```python
from mlxtend.regressor import StackingCVRegressor
from sklearn.linear_model import Lasso
lr = LinearRegression()
rf = RandomForestRegressor(max_depth=5, random_state=1411)
ransac = linear_model.RANSACRegressor(random_state=1411)
svr = svm.SVR()
krr = KernelRidge(alpha=0.1, kernel='rbf', gamma=1.0)
et = ExtraTreesRegressor(random_state=1411)
knn = KNeighborsRegressor(n_neighbors=1)
lasso = Lasso(random_state=1411)

estimators = [lr,rf,ransac,svr,krr,et,knn]
# Training the model
stack = StackingCVRegressor(regressors=estimators,meta_regressor=lasso,
                            random_state=1411,
                            use_features_in_secondary=True)
stack.fit(X_train, y_train)

#Applying the model to make a prediction
y_stack_train_pred = stack.predict(X_train)
y_stack_test_pred = stack.predict(x_test)

# Scoring: 
stack.score(x_test,y_test)
```
